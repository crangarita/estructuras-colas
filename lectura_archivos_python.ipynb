## ✅ Ejercicio 1: Leer un archivo línea por línea
Crea un archivo de texto llamado `ejemplo.txt` con el siguiente contenido:

```
Hola mundo
Bienvenido al curso de Python
Lectura de archivos es divertida
```

Ahora, escribe un código que lea el archivo línea por línea e imprima cada línea en pantalla.

````python
with open('ejemplo.txt', 'r') as archivo:
    for linea in archivo:
        print(linea.strip())
````

## ✅ Ejercicio 2: Contar líneas en un archivo
Escribe un código que cuente cuántas líneas tiene el archivo `ejemplo.txt`.

````python
with open('ejemplo.txt', 'r') as archivo:
    lineas = archivo.readlines()
    print(f"El archivo tiene {len(lineas)} líneas.")
````

## ✅ Ejercicio 3: Contar palabras en un archivo
Escribe un código que cuente cuántas palabras hay en todo el archivo.

````python
with open('ejemplo.txt', 'r') as archivo:
    contenido = archivo.read()
    palabras = contenido.split()
    print(f"El archivo tiene {len(palabras)} palabras.")
````

## ✅ Ejercicio 4: Buscar una palabra en el archivo
Escribe un código que pregunte al usuario una palabra y le diga en qué líneas aparece.

````python
palabra_buscar = input("Ingresa la palabra a buscar: ").lower()
with open('ejemplo.txt', 'r') as archivo:
    for numero_linea, linea in enumerate(archivo, 1):
        if palabra_buscar in linea.lower():
            print(f"La palabra aparece en la línea {numero_linea}: {linea.strip()}")
````

## 📄 Introducción a la librería `csv`

La librería `csv` de Python permite **leer y escribir archivos CSV (Comma-Separated Values)** de manera sencilla. Estos archivos son muy comunes para almacenar datos en forma tabular, como hojas de cálculo exportadas o bases de datos simples.

### 🔧 ¿Por qué usar `csv`?
- Permite leer archivos línea por línea, separando automáticamente los campos por comas (u otro delimitador).
- Facilita trabajar tanto con listas (`csv.reader`) como con diccionarios (`csv.DictReader`), lo que hace que el código sea más claro y organizado.
- También permite escribir archivos CSV fácilmente usando `csv.writer` o `csv.DictWriter`.



## 📚 Concepto rápido: ¿Qué es una lista en Python?

Una **lista** en Python es una colección ordenada de elementos que puede contener números, cadenas, booleanos ¡o incluso otras listas!

### 🔑 Características principales:
- Se define con **corchetes**: `[]`.
- Puede contener elementos de cualquier tipo.
- Es **mutable**, lo que significa que puedes cambiar, agregar o eliminar elementos después de crearla.
- Los elementos tienen un **índice** que comienza en `0`.

## ✅ Ejercicio 5: Leer un archivo CSV sencillo
Crea un archivo llamado `datos.csv` con el siguiente contenido:

```
nombre,edad,ciudad
Juan,25,Bogotá
María,30,Medellín
Pedro,22,Cali
```

Escribe un código que lea este archivo e imprima cada fila como una lista.

````python
import csv

with open('datos.csv', newline='') as archivo_csv:
    lector = csv.reader(archivo_csv)
    for fila in lector:
        print(fila)
````

## 📚 Concepto rápido: ¿Qué es un diccionario en Python?

Un **diccionario** en Python es una colección no ordenada de pares **clave-valor**, donde cada clave está asociada a un valor.

### 🔑 Características principales:
- Se define con **llaves**: `{}`.
- Cada elemento tiene una **clave** y un **valor**, separados por `:`.
- Las **claves** deben ser únicas.
- Es **mutable**, por lo que puedes agregar, cambiar o eliminar elementos después de crearlo.

### 📦 Ejemplo de diccionario
```python
persona = {'nombre': 'Juan', 'edad': 30, 'ciudad': 'Bogotá'}
```


## ✅ Ejercicio 6: Leer un archivo CSV como diccionario
Usa el archivo `datos.csv` y lee cada fila como un diccionario.

````python
import csv

with open('datos.csv', newline='') as archivo_csv:
    lector = csv.DictReader(archivo_csv)
    for fila in lector:
        print(fila)
````

## ✅ Ejercicio 7: Crear una función que lea cualquier archivo
Escribe una función llamada `leer_archivo` que reciba la ruta de un archivo y devuelva su contenido como lista de líneas.

````python
def leer_archivo(ruta):
    with open(ruta, 'r') as archivo:
        return archivo.readlines()

lineas = leer_archivo('ejemplo.txt')
print(lineas)
````

## ✅ Ejercicio 8: Manejo de errores al leer archivos
Modifica la función `leer_archivo` para que maneje el caso en que el archivo no exista, usando try-except.

````python
def leer_archivo_seguro(ruta):
    try:
        with open(ruta, 'r') as archivo:
            return archivo.readlines()
    except FileNotFoundError:
        print(f"Error: El archivo '{ruta}' no existe.")
        return []

lineas = leer_archivo_seguro('archivo_inexistente.txt')
print(lineas)
````
